
# Implementación de una Red Neuronal para la clasificación del dataset Iris

Este notebook implementa una red neuronal multicapa para la clasificación de las especies de flores en el dataset Iris.
Se modulariza el código en funciones para facilitar la comprensión y la ejecución paso a paso.



## Inicialización de pesos y bias

Antes de entrenar la red, se deben inicializar los pesos y bias aleatoriamente.
Para la capa oculta ($ W_0 $, $ b_0 $) y la capa de salida ($ W_1 $, $ b_1 $):

$$ W_0 \in \mathbb{R}^{n_{input} \times n_{hidden}}, \quad b_0 \in \mathbb{R}^{1 \times n_{hidden}} $$
$$ W_1 \in \mathbb{R}^{n_{hidden} \times n_{output}}, \quad b_1 \in \mathbb{R}^{1 \times n_{output}} $$

Los pesos se inicializan de manera aleatoria para romper la simetría en el aprendizaje.


```python

import numpy as np

def initialize_weights(input_size, hidden_size, output_size):
    W0 = np.random.rand(input_size, hidden_size)
    b0 = np.random.rand(1, hidden_size)
    W1 = np.random.rand(hidden_size, output_size)
    b1 = np.random.rand(1, output_size)
    return W0, b0, W1, b1
```


## Propagación hacia adelante

La propagación hacia adelante calcula las activaciones en cada capa de la red.

1. **Capa oculta**:

$$ Z_h = X W_0 + b_0 $$
$$ A_h = \sigma(Z_h) = \frac{1}{1 + e^{-Z_h}} $$

2. **Capa de salida**:

$$ Z_o = A_h W_1 + b_1 $$
$$ A_o = \text{softmax}(Z_o) = \frac{e^{Z_o}}{\sum e^{Z_o}} $$

Donde $ \sigma(x) $ es la función sigmoide y la softmax convierte las salidas en probabilidades.


```python

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def softmax(x):
    expX = np.exp(x - np.max(x, axis=1, keepdims=True))  # Estabilidad numérica
    return expX / np.sum(expX, axis=1, keepdims=True)

def forward_propagation(X, W0, b0, W1, b1):
    Z_h = np.dot(X, W0) + b0
    A_h = sigmoid(Z_h)

    Z_o = np.dot(A_h, W1) + b1
    A_o = softmax(Z_o)

    return A_h, A_o
```


## Retropropagación

El error se propaga hacia atrás para ajustar los pesos mediante el gradiente descendente.

1. **Error en la capa de salida**:

$$ \delta_o = A_o - Y $$

2. **Error en la capa oculta**:

$$ \delta_h = (\delta_o W_1^T) \cdot \sigma'(Z_h) $$

3. **Gradientes para actualizar los pesos**:

$$ W_1 \gets W_1 - \eta (A_h^T \delta_o) $$
$$ b_1 \gets b_1 - \eta \sum \delta_o $$

$$ W_0 \gets W_0 - \eta (X^T \delta_h) $$
$$ b_0 \gets b_0 - \eta \sum \delta_h $$

Donde $ \sigma'(x) $ es la derivada de la función sigmoide.


```python

def sigmoid_deriv(x):
    return x * (1 - x)

def backpropagation(X, Y, A_h, A_o, W1):
    delta_o = A_o - Y

    dcost_dah = np.dot(delta_o, W1.T)
    delta_h = dcost_dah * sigmoid_deriv(A_h)

    return delta_o, delta_h
```


## Actualización de pesos

Se actualizan los pesos usando el algoritmo de gradiente descendente:

$$ W_1 \gets W_1 - \eta (A_h^T \delta_o) $$
$$ b_1 \gets b_1 - \eta \sum \delta_o $$

$$ W_0 \gets W_0 - \eta (X^T \delta_h) $$
$$ b_0 \gets b_0 - \eta \sum \delta_h $$


```python

def update_weights(X, A_h, delta_o, delta_h, W0, b0, W1, b1, learning_rate):
    W1 -= learning_rate * np.dot(A_h.T, delta_o)
    b1 -= learning_rate * np.sum(delta_o, axis=0, keepdims=True)

    W0 -= learning_rate * np.dot(X.T, delta_h)
    b0 -= learning_rate * np.sum(delta_h, axis=0, keepdims=True)

    return W0, b0, W1, b1
```


## Entrenamiento de la red neuronal

El entrenamiento se realiza por múltiples épocas aplicando la propagación hacia adelante, la retropropagación y la actualización de pesos en cada iteración.


```python

def train_neural_network(X_train, Y_train, hidden_size, epochs, learning_rate):
    input_size = X_train.shape[1]
    output_size = Y_train.shape[1]

    W0, b0, W1, b1 = initialize_weights(input_size, hidden_size, output_size)

    for epoch in range(epochs):
        A_h, A_o = forward_propagation(X_train, W0, b0, W1, b1)
        delta_o, delta_h = backpropagation(X_train, Y_train, A_h, A_o, W1)
        W0, b0, W1, b1 = update_weights(X_train, A_h, delta_o, delta_h, W0, b0, W1, b1, learning_rate)

        if epoch % 100 == 0:
            loss = -np.mean(Y_train * np.log(A_o + 1e-9))  # Evitar log(0)
            print(f"Época {epoch}, Pérdida: {loss:.4f}")

    return W0, b0, W1, b1
```
